```python
import numpy as np

# Khởi tạo dữ liệu
w = np.array([-2, 1, 0])
x = np.array([2, 3, 1])
y = 1
eta = 1.0  # Hệ số học (learning rate)

print("=== MỤC 1: KIỂM TRA PHÂN LỚP ===")
# Tính w^T * x
score_before = np.dot(w, x)
print(f"Giá trị w^T * x ban đầu = {score_before}")

# Kiểm tra điều kiện phân lớp sai: y * (w^T * x) <= 0
is_misclassified = (y * score_before) <= 0

if is_misclassified:
    print("-> Mẫu X BỊ PHÂN LỚP SAI!")
else:
    print("-> Mẫu X đã được phân lớp đúng.")

print("\n=== MỤC 2: CẬP NHẬT TRỌNG SỐ PERCEPTRON ===")
if is_misclassified:
    # Công thức: w_new = w + eta * y * x
    w_new = w + eta * y * x
    print(f"Trọng số w mới sau cập nhật = {w_new}")
else:
    w_new = w
    print("Không cần cập nhật trọng số.")

print("\n=== MỤC 3: TÍNH LẠI w^T * x SAU CẬP NHẬT ===")
score_after = np.dot(w_new, x)
print(f"Giá trị w_new^T * x mới = {score_after}")

# Kiểm tra lại kết quả
if (y * score_after) > 0:
    print("-> Sau khi cập nhật, mẫu X đã được PHÂN LỚP ĐÚNG!")
```
