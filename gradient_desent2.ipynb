```python
import numpy as np

def grad(x):
    return x**2 - 1

def cost(x):
    return (1/3)*x**3 - x

def myGD1(x0, eta):
    x = [x0]
    for it in range(100):
        x_new = x[-1] - eta*grad(x[-1])
        if abs(grad(x_new)) < 1e-3:
            break
        if abs(x_new) > 1e6:
            print(f"Xử lý x0 = {x0}: Thuật toán bị phân kỳ ra vô cùng!")
            return (x, it)
        x.append(x_new)
    return (x, it)

(x1, it1) = myGD1(-2.0, .1) 
(x2, it2) = myGD1(2.0, .1) 

print('Trường hợp 1 (x0 = -2): Solution x1 = %f, cost = %f, after %d iterations'%(x1[-1], cost(x1[-1]), it1))
print('Trường hợp 2 (x0 = 2) : Solution x2 = %f, cost = %f, after %d iterations'%(x2[-1], cost(x2[-1]), it2))
```
