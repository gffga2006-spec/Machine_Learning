```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.01, epochs=50, random_state=1):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.random_state = random_state

    def fit(self, X, y):
        """
        Huấn luyện mô hình từ đầu bằng thuật toán Perceptron
        """
        X = np.array(X)
        y = np.array(y)
        
        # Chuyển đổi nhãn về -1 và 1 để phục vụ tính toán
        y_adjusted = np.where(y <= 0, -1, 1)

        # Khởi tạo trọng số ngẫu nhiên bằng NumPy
        rgen = np.random.RandomState(self.random_state)
        self.weights_ = rgen.normal(loc=0.0, scale=0.01, size=X.shape[1])
        self.bias_ = 0.0
        
        self.errors_ = []

        # Vòng lặp huấn luyện qua từng epoch
        for _ in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y_adjusted):
                # Tính giá trị dự báo tạm thời (-1 hoặc 1)
                z = np.dot(xi, self.weights_) + self.bias_
                prediction = 1 if z >= 0.0 else -1
                
                # Cập nhật trọng số theo công thức Perceptron
                update = self.learning_rate * (target - prediction)
                self.weights_ += update * xi
                self.bias_ += update
                
                # Đếm số lỗi
                errors += int(update != 0.0)
                
            self.errors_.append(errors)
            
        return self

    def predict(self, X):
        """
        Dự báo nhãn (0 hoặc 1) cho dữ liệu mới
        """
        X = np.array(X)
        # Tính z = X * w + b
        z = np.dot(X, self.weights_) + self.bias_
        # Trả về 1 nếu z >= 0, ngược lại trả về 0
        return np.where(z >= 0.0, 1, 0)


# ==========================================
# KIỂM THỬ THỬ NGHIỆM (Không dùng sklearn)
# ==========================================

if __name__ == "__main__":
    # Tập dữ liệu học hàm LOGIC AND
    X_train = np.array([
        [0, 0],
        [0, 1],
        [1, 0],
        [1, 1]
    ])
    y_train = np.array([0, 0, 0, 1])

    # Khởi tạo mô hình
    model = Perceptron(learning_rate=0.1, epochs=10, random_state=42)
    
    # Huấn luyện mô hình
    model.fit(X_train, y_train)

    # Dự báo dữ liệu mới
    X_test = np.array([
        [0, 0],
        [1, 1],
        [1, 0],
        [0, 1]
    ])
    predictions = model.predict(X_test)

    print("--- KẾT QUẢ HUẤN LUYỆN ---")
    print("Trọng số (Weights):", model.weights_)
    print("Độ lệch (Bias):    ", model.bias_)
    print("Số lỗi qua từng epoch:", model.errors_)
    print("\n--- KẾT QUẢ DỰ BÁO ---")
    for inputs, pred in zip(X_test, predictions):
        print(f"Đầu vào {inputs} -> Dự báo: {pred}")
```
