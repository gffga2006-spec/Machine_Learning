```python
import numpy as np

# Khai báo các tham số
w = np.array([1, 2, -10])
x = np.array([3, 4, 1])

print("=== BẮT ĐẦU THỰC THI ===")

# -------------------------------------------------------------
# MỤC 1: Tính w^T * x
# -------------------------------------------------------------
w_T_x = np.dot(w, x)
print(f"1. Giá trị w^T * x = {w_T_x}")

# -------------------------------------------------------------
# MỤC 2: Xác định nhãn dự đoán của điểm dữ liệu
# -------------------------------------------------------------
y_pred = 1 if w_T_x >= 0 else -1
print(f"2. Nhãn dự đoán (y_pred) = {y_pred}")

# -------------------------------------------------------------
# MỤC 3: Kiểm tra khi nhãn thực tế y = -1
# -------------------------------------------------------------
y_actual = -1
is_misclassified = (y_pred != y_actual)

print(f"3. Nhãn thực tế (y_actual) = {y_actual}")
if is_misclassified:
    print("   -> ĐÁP ÁN: Điểm dữ liệu BỊ PHÂN LỚP SAI (Misclassified).")
else:
    print("   -> ĐÁP ÁN: Điểm dữ liệu ĐƯỢC PHÂN LỚP ĐÚNG (Correctly classified).")
```
